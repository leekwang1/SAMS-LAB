# SAMSLab 실행 노트북

이 노트북은 현재 커널의 Python 환경으로 `source/PyVITA/main.py`를 실행합니다.
GUI 앱이므로 실행하면 별도 창이 열립니다.

```python
from pathlib import Path
import os
import sys

PROJECT_ROOT = Path.cwd().resolve()
APP_DIR = PROJECT_ROOT / "source" / "PyVITA"
MAIN_FILE = APP_DIR / "main.py"
REQUIREMENTS_FILE = APP_DIR / "requirements.txt"

print(f"Project root: {PROJECT_ROOT}")
print(f"App dir: {APP_DIR}")
print(f"Python: {sys.executable}")

if not MAIN_FILE.exists():
    raise FileNotFoundError(f"실행 파일을 찾지 못했습니다: {MAIN_FILE}")

os.chdir(APP_DIR)
print(f"Working directory changed to: {Path.cwd()}")
```

```python
import importlib

required_modules = {
    "PyQt5": "PyQt5",
    "OpenGL": "PyOpenGL",
    "laspy": "laspy",
    "numpy": "numpy",
    "scipy": "scipy",
    "alphashape": "alphashape",
    "shapely": "shapely",
}

missing = []
for module_name, package_name in required_modules.items():
    try:
        importlib.import_module(module_name)
        print(f"OK: {package_name}")
    except Exception:
        print(f"MISSING: {package_name}")
        missing.append(package_name)

if missing:
    print("\n설치가 필요한 패키지:")
    for package_name in missing:
        print(f"- {package_name}")
    print("\n필요하면 다음 셀의 주석을 해제해 설치하세요.")
else:
    print("\n필수 패키지가 모두 준비되어 있습니다.")
```

```python
# 패키지가 없을 때만 사용하세요.
# %pip install -r requirements.txt
```

```python
import subprocess
import time

if "samlab_process" in globals() and samlab_process.poll() is None:
    print(f"이미 실행 중입니다. PID={samlab_process.pid}")
else:
    env = os.environ.copy()
    env.setdefault("QT_AUTO_SCREEN_SCALE_FACTOR", "1")
    samlab_process = subprocess.Popen(
        [sys.executable, "main.py"],
        cwd=str(APP_DIR),
        env=env,
    )
    time.sleep(1)
    status = "running" if samlab_process.poll() is None else f"exited({samlab_process.returncode})"
    print(f"SAMSLab process started. PID={samlab_process.pid}, status={status}")
    print("앱 창이 열리지 않으면 위 상태와 노트북 출력 메시지를 확인하세요.")
```
